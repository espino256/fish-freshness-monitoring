```python
import os
import hashlib
from collections import defaultdict
from PIL import Image

BASE_PATH = "/content/drive/MyDrive/AquaSense_WholeFish_Dataset/Dataset"

parts = ["EYES", "GILLS", "SCALES"]
classes = ["Fresh", "Moderate", "Spoiled"]

duplicates = defaultdict(list)
corrupted = []
valid_images = 0

for part in parts:
    for category in classes:
        folder = os.path.join(BASE_PATH, part, category)

        for filename in os.listdir(folder):
            if not filename.lower().endswith(
                (".jpg", ".jpeg", ".png")
            ):
                continue

            file_path = os.path.join(folder, filename)

            try:
                with Image.open(file_path) as img:
                    img.verify()

                with open(file_path, "rb") as f:
                    file_hash = hashlib.sha256(f.read()).hexdigest()

                duplicates[file_hash].append(
                    (part, category, file_path)
                )
                valid_images += 1

            except Exception as e:
                corrupted.append((file_path, str(e)))

duplicate_groups = {
    key: paths
    for key, paths in duplicates.items()
    if len(paths) > 1
}

print("Total valid images:", valid_images)
print("Corrupted images:", len(corrupted))
print("Exact duplicate groups:", len(duplicate_groups))
print(
    "Extra exact duplicate files:",
    sum(len(v) - 1 for v in duplicate_groups.values())
)

conflicting_groups = [
    paths for paths in duplicate_groups.values()
    if len({(p, c) for p, c, _ in paths}) > 1
]

print("Cross-category/part duplicate groups:",
      len(conflicting_groups))
```

```python
import os
import shutil
from collections import Counter

CLEAN_PATH = "/content/drive/MyDrive/AquaSense_Clean_Dataset"

# Protect an existing cleaned dataset from being overwritten.
if os.path.exists(CLEAN_PATH):
    raise FileExistsError(
        "Clean dataset folder already exists. "
        "Inspect it before running this code again."
    )

parts = ["EYES", "GILLS", "SCALES"]
classes = ["FRESH", "MODERATE", "SPOILED"]

counts = Counter()
excluded_groups = 0

os.makedirs(CLEAN_PATH)

for file_hash, paths in duplicates.items():

    labels = {
        (part, category)
        for part, category, path in paths
    }

    # Exclude any duplicate group with conflicting labels.
    if len(labels) > 1:
        excluded_groups += 1
        continue

    # Keep one copy of each unique image.
    part, category, source = paths[0]

    destination_folder = os.path.join(
        CLEAN_PATH, part, category
    )

    os.makedirs(destination_folder, exist_ok=True)

    destination = os.path.join(
        destination_folder,
        file_hash + "_" + os.path.basename(source)
    )

    shutil.copy2(source, destination)

    counts[(part, category)] += 1

for part in parts:
    for category in classes:
        os.makedirs(
            os.path.join(CLEAN_PATH, part, category),
            exist_ok=True
        )

print("========== CLEAN DATASET ==========")

for part in parts:
    print("\n", part)
    for category in classes:
        print(category, ":", counts[(part, category)])

print("\nTotal clean images:", sum(counts.values()))
print("Excluded conflicting groups:", excluded_groups)
print("Saved to:", CLEAN_PATH)
```

```python
import os
import imagehash
from PIL import Image
from collections import defaultdict

CLEAN_PATH = "/content/drive/MyDrive/AquaSense_Clean_Dataset"

parts = ["EYES", "GILLS", "SCALES"]
classes = ["FRESH", "MODERATE", "SPOILED"]

hashes = defaultdict(list)

for part in parts:
    for category in classes:
        folder = os.path.join(CLEAN_PATH, part, category)

        for filename in os.listdir(folder):
            if not filename.lower().endswith(
                (".jpg", ".jpeg", ".png")
            ):
                continue

            path = os.path.join(folder, filename)

            with Image.open(path) as img:
                phash = imagehash.phash(img.convert("RGB"))

            hashes[part].append(
                (category, path, phash)
            )

for part in parts:
    images = hashes[part]
    similar_pairs = []

    for i in range(len(images)):
        for j in range(i + 1, len(images)):

            category1, path1, hash1 = images[i]
            category2, path2, hash2 = images[j]

            distance = hash1 - hash2

            if distance <= 5:
                similar_pairs.append(
                    (category1, category2, path1, path2, distance)
                )

    print(f"\n{part}")
    print("Potentially similar image pairs:", len(similar_pairs))
    print(
        "Different-label pairs:",
        sum(a != b for a, b, *_ in similar_pairs)
    )

    for pair in similar_pairs[:5]:
        print(pair)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

for part in parts:
    images = hashes[part]
    suspicious = []

    for i in range(len(images)):
        for j in range(i + 1, len(images)):
            label1, path1, hash1 = images[i]
            label2, path2, hash2 = images[j]

            distance = hash1 - hash2

            if distance <= 5 and label1 != label2:
                suspicious.append(
                    (label1, path1, label2, path2, distance)
                )

    print(f"\n{part}: {len(suspicious)} suspicious pairs")

    for label1, path1, label2, path2, distance in suspicious[:5]:

        fig, axes = plt.subplots(1, 2, figsize=(8, 4))

        for ax, path, label in [
            (axes[0], path1, label1),
            (axes[1], path2, label2)
        ]:
            with Image.open(path) as img:
                ax.imshow(img.convert("RGB"))

            ax.set_title(label)
            ax.axis("off")

        fig.suptitle(
            f"{part} | Perceptual hash distance: {distance}"
        )
        plt.tight_layout()
        plt.show()
```

```python
import os
import shutil
from collections import Counter

CLEAN_PATH = "/content/drive/MyDrive/AquaSense_Clean_Dataset"
FINAL_PATH = "/content/drive/MyDrive/AquaSense_Training_Dataset"
REVIEW_LIST = "/content/drive/MyDrive/AquaSense_Images_For_Review.txt"

parts = ["EYES", "GILLS", "SCALES"]
classes = ["FRESH", "MODERATE", "SPOILED"]

# Read image paths requiring manual review
with open(REVIEW_LIST, "r") as f:
    excluded = {
        os.path.normpath(line.strip())
        for line in f
        if line.strip()
    }

# Prevent accidental overwriting
if os.path.exists(FINAL_PATH):
    raise FileExistsError(
        "Training dataset already exists. "
        "Inspect it before running again."
    )

counts = Counter()
excluded_count = 0

for part in parts:
    for category in classes:
        source_folder = os.path.join(
            CLEAN_PATH, part, category
        )

        destination_folder = os.path.join(
            FINAL_PATH, part, category
        )

        os.makedirs(destination_folder, exist_ok=True)

        for filename in os.listdir(source_folder):
            if not filename.lower().endswith(
                (".jpg", ".jpeg", ".png")
            ):
                continue

            source = os.path.normpath(
                os.path.join(source_folder, filename)
            )

            if source in excluded:
                excluded_count += 1
                continue

            shutil.copy2(
                source,
                os.path.join(destination_folder, filename)
            )

            counts[(part, category)] += 1

print("===== PROVISIONAL TRAINING DATASET =====")

for part in parts:
    print("\n", part)
    for category in classes:
        print(category, ":", counts[(part, category)])

print("\nTotal images:", sum(counts.values()))
print("Excluded for review:", excluded_count)
print("Saved to:", FINAL_PATH)
```

```python
import os
import re
from collections import defaultdict

TRAINING_PATH = "/content/drive/MyDrive/AquaSense_Training_Dataset"

parts = ["EYES", "GILLS", "SCALES"]
classes = ["FRESH", "MODERATE", "SPOILED"]

groups = defaultdict(list)

def extract_group(filename):
    # Remove the SHA-256 prefix added during cleaning
    original = re.sub(r"^[a-fA-F0-9]{64}_", "", filename)

    # Find the camera timestamp
    match = re.search(r"IMG[_-]?(\d{14})", original, re.IGNORECASE)

    if match:
        return match.group(1)

    # Unrecognized filenames remain separate for manual review
    return "UNKNOWN_" + original

for part in parts:
    for category in classes:
        folder = os.path.join(TRAINING_PATH, part, category)

        for filename in os.listdir(folder):
            if filename.lower().endswith((".jpg", ".jpeg", ".png")):
                group_id = extract_group(filename)
                groups[group_id].append(
                    (part, category, filename)
                )

print("Total image groups:", len(groups))

multi_groups = {
    k: v for k, v in groups.items()
    if len(v) > 1
}

print("Groups containing multiple images:", len(multi_groups))

for group_id, images in list(multi_groups.items())[:10]:
    print(group_id, ":", len(images), "images")
```

```python
import os
import re
import csv
import random
from collections import defaultdict, Counter

TRAINING_PATH = "/content/drive/MyDrive/AquaSense_Training_Dataset"
SPLIT_PATH = "/content/drive/MyDrive/AquaSense_Splits.csv"

parts = ["EYES", "GILLS", "SCALES"]
classes = ["FRESH", "MODERATE", "SPOILED"]

random.seed(42)

def get_group(filename):
    original = re.sub(r"^[a-fA-F0-9]{64}_", "", filename)
    match = re.search(r"IMG[_-]?(\d{14})", original, re.I)

    if match:
        return match.group(1)

    return "UNKNOWN_" + original

rows = []

for part in parts:
    for category in classes:

        folder = os.path.join(TRAINING_PATH, part, category)

        for filename in os.listdir(folder):
            if filename.lower().endswith((".jpg", ".jpeg", ".png")):
                rows.append({
                    "part": part,
                    "label": category,
                    "path": os.path.join(folder, filename),
                    "group": get_group(filename)
                })

# Group photos, preserving their timestamp relationships
grouped = defaultdict(list)

for row in rows:
    grouped[row["group"]].append(row)

# Assign groups separately for each class and part
assignments = {}

for part in parts:
    for category in classes:

        keys = [
            key for key, items in grouped.items()
            if all(
                r["part"] == part and r["label"] == category
                for r in items
            )
        ]

        random.shuffle(keys)

        n = len(keys)
        n_train = int(n * 0.70)
        n_val = int(n * 0.15)

        for i, key in enumerate(keys):
            if i < n_train:
                split = "train"
            elif i < n_train + n_val:
                split = "val"
            else:
                split = "test"

            assignments[key] = split

for row in rows:
    row["split"] = assignments.get(row["group"], "REVIEW")

with open(SPLIT_PATH, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=rows[0].keys())
    writer.writeheader()
    writer.writerows(rows)

print("Split file saved:", SPLIT_PATH)

for part in parts:
    print("\n", part)

    for category in classes:
        counts = Counter(
            row["split"]
            for row in rows
            if row["part"] == part and row["label"] == category
        )

        print(category, dict(counts))

assert all(row["split"] != "REVIEW" for row in rows)
assert len(rows) == 2302
print("\nSplit manifest created successfully!")
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

SPLIT_PATH = "/content/drive/MyDrive/AquaSense_Splits.csv"

df = pd.read_csv(SPLIT_PATH)

# Train EYES only
eyes_df = df[df["part"] == "EYES"].copy()

CLASS_NAMES = ["FRESH", "MODERATE", "SPOILED"]
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
SEED = 42

label_map = {
    name: i for i, name in enumerate(CLASS_NAMES)
}

eyes_df["class_id"] = eyes_df["label"].map(label_map)

assert eyes_df["class_id"].notna().all()
assert all(os.path.isfile(p) for p in eyes_df["path"])

print("EYES dataset loaded!")

print(
    eyes_df.groupby(["split", "label"])
    .size()
)
```

```python
import tensorflow as tf

AUTOTUNE = tf.data.AUTOTUNE

def load_image(path, label):
    image = tf.io.read_file(path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    image = tf.cast(image, tf.float32)

    return image, label


def make_dataset(split, training=False):
    subset = eyes_df[eyes_df["split"] == split]

    paths = subset["path"].tolist()
    labels = subset["class_id"].astype("int32").tolist()

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=len(paths),
            seed=SEED
        )

    dataset = dataset.batch(BATCH_SIZE)

    return dataset.prefetch(AUTOTUNE)


train_ds = make_dataset("train", training=True)
val_ds = make_dataset("val")
test_ds = make_dataset("test")

print("Training batches:", len(train_ds))
print("Validation batches:", len(val_ds))
print("Testing batches:", len(test_ds))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

tf.keras.utils.set_random_seed(42)

# Data augmentation
augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10)
], name="augmentation")

# Pretrained MobileNetV2
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers
base_model.trainable = False

# Build classification model
inputs = layers.Input(shape=(224, 224, 3))

x = augmentation(inputs)

# Normalize pixels to MobileNetV2's expected range
x = layers.Rescaling(
    scale=1.0 / 127.5,
    offset=-1.0
)(x)

x = base_model(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    3,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

print("AquaSense EYES model created successfully!")
```

```python
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

train_labels = eyes_df[
    eyes_df["split"] == "train"
]["class_id"].to_numpy()

weights = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(3),
    y=train_labels
)

class_weights = {
    i: float(weight)
    for i, weight in enumerate(weights)
}

print("AquaSense EYES class weights:")
print(class_weights)
```

```python
import tensorflow as tf

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    class_weight=class_weights,
    callbacks=callbacks
)

print("AquaSense EYES model training completed!")
```

```python
import os
from collections import defaultdict

SEARCH_FOLDERS = [
    "/content/drive/MyDrive/AquaSense_Training_Dataset",
    "/content/drive/MyDrive/AquaSense_Clean_Dataset",
    "/content/drive/MyDrive/AquaSense_WholeFish_Dataset"
]

missing_names = {
    os.path.basename(path)
    for path in missing_files
}

found_files = defaultdict(list)

for base in SEARCH_FOLDERS:

    if not os.path.isdir(base):
        continue

    for root, dirs, files in os.walk(base):

        for filename in files:

            if filename in missing_names:
                found_files[filename].append(
                    os.path.join(root, filename)
                )

print("===== AQUASENSE FILE SEARCH =====")

print("Missing image paths:", len(missing_files))
print("Filenames found elsewhere:", len(found_files))

print(
    "Still not located:",
    len(missing_names - set(found_files))
)

print("\nSample matches:")

for filename, paths in list(found_files.items())[:5]:
    print("\nFilename:", filename)

    for path in paths:
        print("FOUND:", path)
```

```python
import os
import pandas as pd

SPLIT_PATH = "/content/drive/MyDrive/AquaSense_Splits.csv"
FIXED_PATH = "/content/drive/MyDrive/AquaSense_Splits_Fixed.csv"

df = pd.read_csv(SPLIT_PATH)
repaired = 0

# Repair paths using previously located filenames.
for idx, row in df.iterrows():
    old_path = row["path"]

    if os.path.isfile(old_path):
        continue

    filename = os.path.basename(old_path)
    candidates = found_files.get(filename, [])

    # Only accept the same body part and class.
    valid = [
        p for p in candidates
        if os.path.isfile(p)
        and os.path.basename(os.path.dirname(p)) == row["label"]
        and os.path.basename(
            os.path.dirname(os.path.dirname(p))
        ) == row["part"]
    ]

    if len(valid) == 1:
        df.at[idx, "path"] = valid[0]
        repaired += 1

print("Repaired paths:", repaired)

# Verify that all EYES test images can be found.
eyes_test = df[
    (df["part"] == "EYES") &
    (df["split"] == "test")
]

remaining = [
    p for p in eyes_test["path"]
    if not os.path.isfile(p)
]

print("EYES test images:", len(eyes_test))
print("EYES test paths still missing:", len(remaining))

assert len(eyes_test) == 143
assert len(remaining) == 0, "Some test files still need repair."

df.to_csv(FIXED_PATH, index=False)

print("Corrected split manifest saved:", FIXED_PATH)
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

CLASS_NAMES = ["FRESH", "MODERATE", "SPOILED"]
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
SEED = 42
AUTOTUNE = tf.data.AUTOTUNE

label_map = {
    "FRESH": 0,
    "MODERATE": 1,
    "SPOILED": 2
}

gills_df["class_id"] = gills_df["label"].map(label_map)

assert gills_df["class_id"].notna().all()

def load_gills_image(path, label):
    image = tf.io.read_file(path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)

    return image, label


def make_gills_dataset(split, training=False):
    subset = gills_df[
        gills_df["split"] == split
    ]

    paths = subset["path"].tolist()
    labels = subset["class_id"].astype("int32").tolist()

    ds = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    ds = ds.map(
        load_gills_image,
        num_parallel_calls=AUTOTUNE
    )

    if training:
        ds = ds.shuffle(
            buffer_size=len(paths),
            seed=SEED
        )

    ds = ds.batch(BATCH_SIZE)

    return ds.prefetch(AUTOTUNE)


gills_train_ds = make_gills_dataset("train", True)
gills_val_ds = make_gills_dataset("val")
gills_test_ds = make_gills_dataset("test")

print("GILLS datasets prepared!")
print("Training batches:", len(gills_train_ds))
print("Validation batches:", len(gills_val_ds))
print("Testing batches:", len(gills_test_ds))
```

```python
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2

tf.keras.utils.set_random_seed(SEED)

gills_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10),
], name="gills_augmentation")

gills_base = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

gills_base.trainable = False

inputs = layers.Input(shape=(224, 224, 3))

x = gills_augmentation(inputs)

x = layers.Rescaling(
    scale=1.0 / 127.5,
    offset=-1.0
)(x)

x = gills_base(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    3,
    activation="softmax"
)(x)

gills_model = Model(inputs, outputs)

gills_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

gills_model.summary()

print("AquaSense GILLS model created successfully!")
```

```python
from sklearn.utils.class_weight import compute_class_weight

gills_train_labels = gills_df[
    gills_df["split"] == "train"
]["class_id"].to_numpy()

weights = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(3),
    y=gills_train_labels
)

gills_class_weights = {
    i: float(weight)
    for i, weight in enumerate(weights)
}

print("GILLS class weights:", gills_class_weights)
```

```python
gills_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]

gills_history = gills_model.fit(
    gills_train_ds,
    validation_data=gills_val_ds,
    epochs=20,
    class_weight=gills_class_weights,
    callbacks=gills_callbacks
)

print("AquaSense GILLS training completed!")
```

```python
MODEL_DIR = "/content/drive/MyDrive/AquaSense_Trained_Models"

os.makedirs(MODEL_DIR, exist_ok=True)

GILLS_MODEL_PATH = os.path.join(
    MODEL_DIR,
    "aquasense_gills_model.keras"
)

gills_model.save(GILLS_MODEL_PATH)

print("GILLS model saved:", GILLS_MODEL_PATH)
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

CLASS_NAMES = ["FRESH", "MODERATE", "SPOILED"]

# Reload the corrected split manifest
SPLIT_PATH = (
    "/content/drive/MyDrive/"
    "AquaSense_Splits_Fixed.csv"
)

df = pd.read_csv(SPLIT_PATH)

test_rows = df[
    (df["part"] == "GILLS") &
    (df["split"] == "test")
].copy()

label_map = {
    "FRESH": 0,
    "MODERATE": 1,
    "SPOILED": 2
}

test_rows["class_id"] = test_rows["label"].map(label_map)

# Check paths before evaluation
missing = [
    p for p in test_rows["path"]
    if not os.path.isfile(p)
]

print("GILLS test images:", len(test_rows))
print("Missing test images:", len(missing))

assert len(test_rows) == 133
assert len(missing) == 0, "Missing test images detected!"

# Prepare input images
def load_test_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, (224, 224))
    image = tf.cast(image, tf.float32)

    return image, label

test_ds = tf.data.Dataset.from_tensor_slices((
    test_rows["path"].tolist(),
    test_rows["class_id"].astype("int32").tolist()
))

test_ds = test_ds.map(load_test_image)
test_ds = test_ds.batch(16)

# Load trained model
MODEL_PATH = (
    "/content/drive/MyDrive/"
    "AquaSense_Trained_Models/"
    "aquasense_gills_model.keras"
)

model = tf.keras.models.load_model(MODEL_PATH)

# Generate predictions
y_true = test_rows["class_id"].to_numpy()

y_prob = model.predict(test_ds)
y_pred = np.argmax(y_prob, axis=1)

print("\n===== AQUASENSE GILLS TEST RESULTS =====")
print(
    f"Test accuracy: "
    f"{accuracy_score(y_true, y_pred) * 100:.2f}%"
)

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        digits=4,
        zero_division=0
    )
)

cm = confusion_matrix(y_true, y_pred, labels=[0, 1, 2])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=CLASS_NAMES
)

disp.plot(cmap="Blues", values_format="d")
plt.title("AquaSense GILLS Confusion Matrix")
plt.show()
```

```python
import os
import pandas as pd

SPLIT_PATH = "/content/drive/MyDrive/AquaSense_Splits_Fixed.csv"

df = pd.read_csv(SPLIT_PATH)

scales_df = df[df["part"] == "SCALES"].copy()

missing = [
    p for p in scales_df["path"]
    if not os.path.isfile(p)
]

print("===== AQUASENSE SCALES DATASET =====")
print("Total images:", len(scales_df))
print("Missing image paths:", len(missing))

print("\nSplit distribution:")
print(
    pd.crosstab(
        scales_df["label"],
        scales_df["split"]
    )
)

print("\nMissing file examples:")
for path in missing[:5]:
    print(path)
```

```python
import tensorflow as tf

CLASS_NAMES = ["FRESH", "MODERATE", "SPOILED"]
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
SEED = 42

label_map = {
    "FRESH": 0,
    "MODERATE": 1,
    "SPOILED": 2
}

scales_df["class_id"] = scales_df["label"].map(label_map)

def load_scales_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)

    return image, label

def make_scales_dataset(split, training=False):
    subset = scales_df[scales_df["split"] == split]

    paths = subset["path"].tolist()
    labels = subset["class_id"].astype("int32").tolist()

    ds = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    ds = ds.map(
        load_scales_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if training:
        ds = ds.shuffle(len(paths), seed=SEED)

    return ds.batch(BATCH_SIZE).prefetch(
        tf.data.AUTOTUNE
    )

scales_train_ds = make_scales_dataset("train", True)
scales_val_ds = make_scales_dataset("val")
scales_test_ds = make_scales_dataset("test")

print("SCALES datasets prepared!")
print("Training batches:", len(scales_train_ds))
print("Validation batches:", len(scales_val_ds))
print("Testing batches:", len(scales_test_ds))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2

tf.keras.utils.set_random_seed(SEED)

# Image augmentation for training
scales_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.10)
], name="scales_augmentation")

# Pretrained MobileNetV2
scales_base = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers
scales_base.trainable = False

# Build SCALES classifier
inputs = layers.Input(shape=(224, 224, 3))

x = scales_augmentation(inputs)

# Normalize pixels from 0–255 to -1 to 1
x = layers.Rescaling(
    scale=1.0 / 127.5,
    offset=-1.0
)(x)

x = scales_base(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    3,
    activation="softmax"
)(x)

scales_model = Model(inputs, outputs)

scales_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

scales_model.summary()

print("AquaSense SCALES model created successfully!")
```

```python
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

scales_train_labels = scales_df[
    scales_df["split"] == "train"
]["class_id"].to_numpy()

weights = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(3),
    y=scales_train_labels
)

scales_class_weights = {
    i: float(weight)
    for i, weight in enumerate(weights)
}

print("SCALES class weights:")
print(scales_class_weights)
```

```python
import os

MODEL_DIR = "/content/drive/MyDrive/AquaSense_Trained_Models"
os.makedirs(MODEL_DIR, exist_ok=True)

SCALES_MODEL_PATH = os.path.join(
    MODEL_DIR,
    "aquasense_scales_model.keras"
)

scales_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ModelCheckpoint(
        filepath=SCALES_MODEL_PATH,
        monitor="val_loss",
        save_best_only=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]

scales_history = scales_model.fit(
    scales_train_ds,
    validation_data=scales_val_ds,
    epochs=20,
    class_weight=scales_class_weights,
    callbacks=scales_callbacks
)

print("AquaSense SCALES training completed!")
print("Saved model:", SCALES_MODEL_PATH)
```
